```python
import pandas as pd
import numpy as np
from pathlib import Path

import plotly.express as px
pd.options.plotting.backend = 'plotly'
```

```python
recipes = pd.read_csv("data/RAW_recipes.csv")
ratings = pd.read_csv("data/RAW_interactions.csv")
```

```python
ratings
```

```python
# Merge dataframes
merged = recipes.merge(ratings, how="left", left_on="id", right_on="recipe_id")
merged['rating'] = merged['rating'].replace(0, np.nan)
```

```python
merged
```

```python
# Find average rating per recipe and add to recipes dataframe
avg_ratings = merged.groupby('id')['rating'].mean()
avg_ratings_df = avg_ratings.reset_index().rename(columns={'rating': 'avg_rating'})

recipes = recipes.merge(avg_ratings_df, how='left', on='id')
```

```python
# Finding columns with missing vals
nan_columns = recipes.columns[recipes.isnull().any()]

print(nan_columns)
```

```python
# Fill missing descriptions with a placeholder
recipes['description'] = recipes['description'].fillna('No description available')

print(f"Number of missing descriptions after filling: {recipes['description'].isna().sum()}")
```

```python
# Check for missing columns
nan_columns = recipes.columns[recipes.isnull().any()]

print(nan_columns)
```

```python
# Check missing values in Average Rating
recipes[recipes['avg_rating'].isna()]
```

We would try to fill in missingness here, but these recipes have no ratings, and such cannot be predicted.

```python
# Convert nutrition column from string to list
recipes['nutrition'] = recipes['nutrition'].apply(eval)
```

```python
# Convert nutrition list into separate columns
nutrition_df = pd.DataFrame(recipes['nutrition'].tolist(), 
                          columns=['calories', 'total fat (PDV)', 'sugar (PDV)', 'sodium (PDV)', 'protein (PDV)', 'saturated fat (PDV)', 'carbohydrates (PDV)'])

# Add the nutrition columns to the original dataframe
recipes = pd.concat([recipes, nutrition_df], axis=1)
```

```python
recipes
```

```python
# Convert times to time series
recipes['submitted'] = pd.to_datetime(recipes['submitted'])
```

```python
# Head of cleaned dataframe
recipes.head()
```

```python
# Distribution of average ratings per recipe
fig = px.histogram(
    recipes,
    x='avg_rating',
    nbins=20,
    title='Distribution of Average Recipe Ratings',
    labels={'avg_rating': 'Average Rating'},
    template='plotly_white'
)
fig.update_layout(title_x=0.5)
fig.show()
```

```python
ratings_count = ratings.groupby('recipe_id').size().rename('review_count').to_frame()

ratings_and_count = avg_ratings_df.set_index('id').dropna().merge(ratings_count, left_on='id', right_on='recipe_id').sort_values('review_count', ascending=False)
ratings_and_count.head(10)
```

```python
# apply bayesian average ratings

global_average = ratings_and_count['avg_rating'].mean()
minimum_reviews = 100 # tune this parameter

ratings_and_count['weighted_rating'] = (
    (ratings_and_count['review_count'] / (ratings_and_count['review_count'] + minimum_reviews)) 
        * ratings_and_count['avg_rating']
    +
    (minimum_reviews / (ratings_and_count['review_count'] + minimum_reviews)) 
        * global_average
)

ratings_and_count.sort_values('weighted_rating', ascending=False)
```

```python
fig = px.scatter(
    ratings_and_count[ratings_and_count['review_count'] >= 100],
    x='review_count',
    y='avg_rating',
    color='avg_rating',
    log_x=True 
)
fig.update_layout(title='Average Rating vs. Number of Reviews')
fig.show()
```
